```python
import pandas as pd

from nba_api.stats.endpoints import playergamelog
from nba_api.stats.static import players

import warnings

# Suppress all warnings
warnings.filterwarnings("ignore")
```

```python
# Hardcoded list of known 70+ point games

recent_high_scoring_games = [
    {"PLAYER_NAME": "Kobe Bryant", "PLAYER_ID": 977, "GAME_DATE": "2006-01-22"},
    {"PLAYER_NAME": "Devin Booker", "PLAYER_ID": 1626164, "GAME_DATE": "2017-03-24"},
    {"PLAYER_NAME": "Donovan Mitchell", "PLAYER_ID": 1628378, "GAME_DATE": "2023-01-02"},
    {"PLAYER_NAME": "Damian Lillard", "PLAYER_ID": 203081, "GAME_DATE": "2023-02-26"},
    {"PLAYER_NAME": "Joel Embiid", "PLAYER_ID": 203954, "GAME_DATE": "2024-01-22"},
    {"PLAYER_NAME": "Luka Dončić", "PLAYER_ID": 1629029, "GAME_DATE": "2024-01-26"},
    {"PLAYER_NAME": "Bam Adebayo", "PLAYER_ID": 1628389, "GAME_DATE": "2026-03-10"}
]
```

## ***Pull these 70 Point Games from the API into 1 DataFrame***

```python
seasons = ['2005-06', '2016-17', '2022-23','2022-23', '2023-24', '2023-24', '2025-26']

date = ["Jan 22, 2006", "Mar 24, 2017", "Jan 02, 2023", "Feb 26, 2023", "Jan 22, 2024", "Jan 26, 2024", "Mar 10, 2026"]

all_dfs = []

for game in recent_high_scoring_games:  # devin and donovan missing??
    gamelog = playergamelog.PlayerGameLog(player_id=game['PLAYER_ID'], 
                            season = seasons[recent_high_scoring_games.index(game)], 
                            season_type_all_star = 'Regular Season')

    df = gamelog.get_data_frames()[0]

    df = df[df["GAME_DATE"] == date[recent_high_scoring_games.index(game)]]
    df['GAME_DATE'] = game["GAME_DATE"]
    df['Player'] = game['PLAYER_NAME']

    all_dfs.append(df)

all_dfs = pd.concat(all_dfs, ignore_index=True)

all_dfs
```

# Since the database only goes back to 1999, we need to import the older 70 point games recorded

## ***Import the Historical Games.csv Dataset and concatenate it with the above DataFrame***

```python
dfold = pd.read_csv("Historical Games.csv")

dfold
```

```python
# Combine into one DataFrame

all_stats = pd.concat([all_dfs, dfold], ignore_index=True)

all_stats
```

## ***Create a "TEAM" column by using string manipulation on the "MATCHUP" column***

```python
all_stats['TEAM'] = all_stats['MATCHUP'].str[:3]

all_stats
```

## ***Given the average number of possessions per decade, create a PTS (NORMALIZED) column to account for the different time peroids the players played in***

```python
# The following is the average number of possessions per decade

possessions_by_decade = {1950: 93, 
                         1960: 115,
                         1970: 106, 
                         1980: 103, 
                         1990: 94,
                         2000: 91,
                         2010: 96, 
                         2020: 99}
```

```python
# Normalize points by possessions

for i in all_stats:
    all_stats['DECADE'] = (all_stats['GAME_DATE'].str[:4].astype(int) // 10 * 10)
    all_stats['POSSESSIONS'] = all_stats['DECADE'].map(possessions_by_decade)
    all_stats['PTS(Normalized)'] = 100 * (all_stats['PTS'].astype(int) / all_stats['POSSESSIONS'])

all_stats 
```

## ***Sort the DataFrame by the PTS(Normalized) column***

```python
import polars as pl
import janitor.polars
import polars.selectors as cs
import great_tables as gt
import gt_extras as gte
```

```python
df = pl.from_pandas(all_stats)
```

```python
df = (
    df
    .sort("PTS(Normalized)", descending=True)
    .select([
          "TEAM","Player", "GAME_DATE", "WL", "PTS", "PTS(Normalized)"
      ])
)
```

## ***Create a Great Tables visualization of your findings***

```python
# file path for team logos
img_paths = "https://raw.githubusercontent.com/Henryjean/data/refs/heads/main/square_nba_logos/"

# Generate table
table = (
    gt.GT(df)
    .tab_header(
        title=gt.md("**Greatest NBA Scoring of All Time**"),
        subtitle="Normalized Scoring for Possessions on Average per Decade",
    )
    .cols_label(**{
        "TEAM": "TEAM",
        "Player": "PLAYER", 
        "GAME_DATE": "GAME_DATE",
        "WL": "WL", 
        "PTS": "PTS",
        "PTS(Normalized)": "PTS (Normalized)"
    })

    .fmt_number(columns="PTS(Normalized)", decimals=0)
    .fmt_image(columns = "TEAM", path=img_paths, file_pattern='{}.svg')
    .cols_align(columns = ["PTS"], align = 'right')
    .cols_align(columns = ["PTS(Normalized)"], align = 'center')
    .data_color(
        columns = "PTS(Normalized)",
        domain=[60, 90],
        palette=["orange", "lightgreen", "blue", "purple"],
        na_color="white",
    )
    .tab_options(
        container_width = "100%",
        table_background_color="white",
        row_group_background_color="white",
        table_font_names="Roboto",
        column_labels_font_size="14px",
        column_labels_font_weight="bold", 
        data_row_padding='0px', 
        table_body_hlines_color='transparent', 
        column_labels_border_top_color='black', 
        column_labels_border_top_width="1px",
        column_labels_border_bottom_style='none', 
        heading_border_bottom_style = "none",
        table_body_border_bottom_style="none",
        table_body_border_bottom_color = "white",
        table_border_bottom_style = 'none',
        table_border_top_style = 'none',
        source_notes_border_lr_style = "none"
    )
)
```
